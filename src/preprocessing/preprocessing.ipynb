```python
# Real Estate Forecasting - Preprocessing
# This notebook is responsible for preprocessing the data ingested in the previous step. 
# The data is cleaned and transformed to be ready for the next steps in the pipeline. 
# The cleaned data is saved as a pickle file for later use in the pipeline.

# Importing necessary libraries
import pandas as pd
import numpy as np
from pathlib import Path
import unicodedata
```

```python
# Importing the data from the CSV file
BASE_DIR = Path.cwd().parent.parent
data_path = BASE_DIR / "data" / "real_estate_data.csv"
df = pd.read_csv(data_path, encoding="utf-8")
df.sample(5)
```

```python
df.describe()
```

```python
# Dropping the unnamed column which is not needed for analysis
df.drop(columns=["Unnamed: 16"], inplace=True)
```

```python
# Converting the 'ano' column to integer type, as it represents the year and should be an integer
df["ano"] = df["ano"].astype("int64")
```

```python
# Defining the identifier columns that will be used for grouping and analysis
id_cols = [
    "ano",
    "trim",
    "geo",
    "cod-ca",
    "ca",
    "cod-prv",
    "prv",
    "sector"
]

# Reshaping the DataFrame from wide format to long format using the melt function, which is useful for analysis and visualization
df_long = df.melt(
    id_vars=id_cols,
    var_name="variable",
    value_name="valor"
)
```

```python
df_long.sample(5)
```

```python
df_long = df_long.dropna(subset=["valor"])
```

```python
# Splitting the 'variable' column into two separate columns: 'tipo' and 'metrica', based on the delimiter '-'
df_long[["tipo", "metrica"]] = (
    df_long["variable"]
    .str.split("-", expand=True)
)
```

```python
# Parsing the 'valor' column to handle mixed number formats, such as European and English formats, and converting them to a consistent float format
def parse_mixed_number(x):
    if pd.isna(x):
        return np.nan

    x = str(x).strip()

    # Caso 1: formato europeo "465,34"
    if "," in x and "." not in x:
        return float(x.replace(",", "."))

    # Caso 2: formato inglés "175.0"
    if "." in x and "," not in x:
        return float(x)

    # Caso 3: miles + decimal "1.234,56"
    if "." in x and "," in x:
        x = x.replace(".", "").replace(",", ".")
        return float(x)

    # Caso 4: solo número entero
    try:
        return float(x)
    except:
        return np.nan
```

```python
# Applying the parsing function to the 'valor' column to convert all mixed number formats to a consistent float format for analysis
df_long["valor"] = df_long["valor"].apply(parse_mixed_number)
```

```python
# Selecting only the columns with object data type for further analysis or processing
df_object = df_long.select_dtypes(include=['object'])

# Iterating through each column in the DataFrame that contains object data type and applying string operations to clean the data 
# by stripping leading and trailing whitespace and converting all text to lowercase for consistency
for col in df_object.columns:
    df_long[col] = df_long[col].str.strip()
    df_long[col] = df_long[col].str.lower()
```

```python
# Removing accents from the text to standardize the data and avoid issues with encoding or comparisons
def remove_accents(text):
    if isinstance(text, str):
        text = unicodedata.normalize("NFKD", text)
        text = "".join([c for c in text if not unicodedata.combining(c)])
    return text

for col in df_object.columns:
    df_long[col] = df_long[col].apply(remove_accents)
```

```python
# Saving the cleaned DataFrame as a pickle file for later use in the pipeline
output_path = BASE_DIR / "data" / "cleaned_real_estate_data.pkl"
df_long.to_pickle(output_path)
```
