```python
# Real Estate Forecasting - Ingestion
# This notebook is responsible for ingesting the data from the source and preparing it for the next steps in the pipeline.
# The data is ingested from a CSV file and stored in a pandas DataFrame. The DataFrame is then saved as a pickle file for later use in the pipeline.

# Importing necessary libraries
import pandas as pd
import numpy as np
import os
from pathlib import Path
```

```python
# Ingesting the data from the CSV files

BASE_DIR = Path.cwd().parent.parent

residential_path = BASE_DIR / "data" / "RP_ComprvResid_2007-4t2025.csv"
commercial_path = BASE_DIR / "data" / "RP_ComprvCom_2007-4t2025.csv"

db_residential = pd.read_csv(residential_path, sep=";", encoding="utf-8")
db_commercial = pd.read_csv(commercial_path, sep=";", encoding="utf-8")
```

```python
db_residential.sample(5)
```

```python
db_commercial.sample(5)
```

```python
# Adding a new column to identify the sector of the data
db_residential['sector'] = 'residential'
db_commercial['sector'] = 'commercial'

# Concatenating the two DataFrames into one
db = pd.concat([db_residential, db_commercial], axis = 0, ignore_index=True)
```

```python
db.sample(5)
```

```python
# Saving the new DataFrame concatenated
db.to_csv(BASE_DIR / "data" / "real_estate_data.csv", index=False, encoding="utf-8")
```
